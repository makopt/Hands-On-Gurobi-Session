# Solving the 1-Dimensional Bin Packing Problem Using Gurobi in Python

The **Bin Packing Problem** is a classic optimization problem where the goal is to **pack items of different volumes into the minimum number of bins** of fixed capacity. 

In this tutorial, we will use **Gurobi**, a powerful optimization solver, to solve the 1D Bin Packing Problem in **Python**.

## Problem Definition
Given a **set $N$ of $n$ items** $N=\{1, 2, \dots, n\}$, each with a **volume $v_i$**, and a **set $M$  of $m$ bins** $M=\{1, 2, \dots, m\}$, each with a **capacity $C_j$**. The **bin packing problem** aims to determine the **minimum number of bins** needed to pack all items without exceeding the capacity of any bin.

---

## Standard Mathematical Formulation
The 1-Dimensional Bin Packing Problem can be formulated as follows:

- **Decision Variables**:
  - $x_{ij} = 1$ if item $i$ is packed in bin $j$, $0$ otherwise
  - $y_j = 1$ if bin $j$ is used, $0$ otherwise

- **Objective Function**:
  - Min Z = $\sum_{j=1}^{m} y_j$ (minimize number of bins used)

- **Constraints**:
  - $\sum_{j=1}^{m} x_{ij} = 1$ $\forall i \in N$ (each item is packed in exactly one bin)
  - $\sum_{i=1}^{n} v_i x_{ij} \leq C_j y_j$ $\forall j \in M$ (bin capacity constraint)

## First Example (Solving a small manually created instance of the Bin Packing Problem)

Let's solve a small manually created instance of the Bin Packing Problem using Gurobi in Python.

### Overview

This Python code is a straightforward implementation for solving a **1-Dimensional Bin Packing Problem** using the Gurobi optimizer.

### Key Characteristics

*   **Manual Instance:** The problem data (volumes for 10 items and bin capacity) is hard-coded.
*   **Basic Implementation:** Demonstrates the core steps of setting up and solving a MIP model with Gurobi.
*   **Visualization Included:** The solution is presented both textually and through a detailed matplotlib visualization.

### Code Implementation

```python
from gurobipy import GRB, Model
import numpy as np
import matplotlib.pyplot as plt

# Prepare data
n = 10  # number of items
m = 4   # number of bins (upper bound)
volumes = [13, 6, 9, 10, 8, 7, 13, 12, 10, 7]  # volumes of items
capacities = [65, 30, 60, 50]  # capacities of each bin

# Create a new model
model = Model("bin-packing")

# Create variables
x = model.addVars(n, m, vtype=GRB.BINARY, name="x")
y = model.addVars(m, vtype=GRB.BINARY, name="y")

# Set objective: minimize number of bins used
model.setObjective(y[0] + y[1] + y[2] + y[3], GRB.MINIMIZE)

# Add constraints
# Each item must be in exactly one bin
model.addConstr(x[0, 0] + x[0, 1] + x[0, 2] + x[0, 3] == 1, f"item_{0}")
model.addConstr(x[1, 0] + x[1, 1] + x[1, 2] + x[1, 3] == 1, f"item_{1}")
model.addConstr(x[2, 0] + x[2, 1] + x[2, 2] + x[2, 3] == 1, f"item_{2}")
model.addConstr(x[3, 0] + x[3, 1] + x[3, 2] + x[3, 3] == 1, f"item_{3}")
model.addConstr(x[4, 0] + x[4, 1] + x[4, 2] + x[4, 3] == 1, f"item_{4}")
model.addConstr(x[5, 0] + x[5, 1] + x[5, 2] + x[5, 3] == 1, f"item_{5}")
model.addConstr(x[6, 0] + x[6, 1] + x[6, 2] + x[6, 3] == 1, f"item_{6}")
model.addConstr(x[7, 0] + x[7, 1] + x[7, 2] + x[7, 3] == 1, f"item_{7}")
model.addConstr(x[8, 0] + x[8, 1] + x[8, 2] + x[8, 3] == 1, f"item_{8}")
model.addConstr(x[9, 0] + x[9, 1] + x[9, 2] + x[9, 3] == 1, f"item_{9}")

# Bin capacity constraints
model.addConstr(volumes[0]*x[0, 0] + volumes[1]*x[1, 0] + volumes[2]*x[2, 0] + volumes[3]*x[3, 0] + volumes[4]*x[4, 0] + volumes[5]*x[5, 0] + volumes[6]*x[6, 0] + volumes[7]*x[7, 0] + volumes[8]*x[8, 0] + volumes[9]*x[9, 0] <= capacities[0]*y[0], "capacity_0")
model.addConstr(volumes[0]*x[0, 1] + volumes[1]*x[1, 1] + volumes[2]*x[2, 1] + volumes[3]*x[3, 1] + volumes[4]*x[4, 1] + volumes[5]*x[5, 1] + volumes[6]*x[6, 1] + volumes[7]*x[7, 1] + volumes[8]*x[8, 1] + volumes[9]*x[9, 1] <= capacities[1]*y[1], "capacity_1")
model.addConstr(volumes[0]*x[0, 2] + volumes[1]*x[1, 2] + volumes[2]*x[2, 2] + volumes[3]*x[3, 2] + volumes[4]*x[4, 2] + volumes[5]*x[5, 2] + volumes[6]*x[6, 2] + volumes[7]*x[7, 2] + volumes[8]*x[8, 2] + volumes[9]*x[9, 2] <= capacities[2]*y[2], "capacity_2")
model.addConstr(volumes[0]*x[0, 3] + volumes[1]*x[1, 3] + volumes[2]*x[2, 3] + volumes[3]*x[3, 3] + volumes[4]*x[4, 3] + volumes[5]*x[5, 3] + volumes[6]*x[6, 3] + volumes[7]*x[7, 3] + volumes[8]*x[8, 3] + volumes[9]*x[9, 3] <= capacities[3]*y[3], "capacity_3")

# write model to file
model.write("small_binpacking.lp")

# Optimize model
model.optimize()

# Print solution
print(f"Optimal number of bins: {model.objVal}")
for j in range(m):
    if y[j].X > 0.5:
        bin_items = [i for i in range(n) if x[i, j].X > 0.5]
        bin_volume = sum(volumes[i] for i in bin_items)
        print(f"Bin {j+1}: Items {[i+1 for i in bin_items]}, Total volume: {bin_volume}/{capacities[j]}")

# Visualization
plt.figure(figsize=(12, 8))
colors = plt.cm.Set3(np.linspace(0, 1, m))

for j in range(m):
    if y[j].X > 0.5:
        bin_items = [i for i in range(n) if x[i, j].X > 0.5]
        bin_volumes = [volumes[i] for i in bin_items]
        
        # Create stacked bar for this bin
        bottom = 0
        for i, vol in enumerate(bin_volumes):
            plt.bar(j, vol, bottom=bottom, color=colors[j], alpha=0.7)
            plt.text(j, bottom + vol/2, f"Item {bin_items[i]+1}\n({vol})", 
                    ha='center', va='center', fontsize=8)
            bottom += vol
    # Draw bin capacity line
    plt.hlines(capacities[j], j-0.4, j+0.4, colors='red', linestyles='dashed')
    plt.text(j, capacities[j]+1, f"Capacity: {capacities[j]}", 
                ha='center', va='bottom', color='red', fontsize=8)
plt.xlabel('Bin Number')
plt.ylabel('Volume')
plt.title('Bin Packing Solution')
plt.legend()
plt.xticks(range(m), [f'Bin {i+1}' for i in range(m)])
plt.show()
```

# Second Example (Solving random generated instances of the Bin Packing Problem)

## Overview

This Python code is a comprehensive benchmarking suite for the 1D Bin Packing Problem that systematically tests Gurobi's performance on randomly generated instances.



```python
from gurobipy import quicksum
import openpyxl as pxl
import random
import os

# Function to generate bin packing instances
def generate_binpacking_instance(num_items, num_bins, seed, capacity_ratio):
    random.seed(seed)
    np.random.seed(seed)
    
    # Generate item volumes (mostly small items with some large ones)
    volumes = np.random.randint(1, 50, size=num_items)
    
    # Add a few large items
    large_items = np.random.choice(num_items, size=max(1, num_items // 10), replace=False)
    for idx in large_items:
        volumes[idx] = np.random.randint(30, 70)

    # Set bins capacities (based on average volume and capacity ratio)
    avg_volume = np.mean(volumes)
    capacities = [random.randint(int(avg_volume * capacity_ratio), int(avg_volume * capacity_ratio * 2)) for _ in range(num_bins)]

    return volumes, capacities

# Function to solve the bin packing problem
def solve_binpacking(volumes, capacities, display):
    n = len(volumes)
    m = len(capacities)
    
    model = Model("random-bin-packing")
    
    if not display:
        model.setParam('OutputFlag', 0)
    
    # Create variables
    x = model.addVars(n, m, vtype=GRB.BINARY, name="x")
    y = model.addVars(m, vtype=GRB.BINARY, name="y")
    
    # Set objective
    model.setObjective(quicksum(y[j] for j in range(m)), GRB.MINIMIZE)
    
    # Add constraints
    for i in range(n):
        model.addConstr(quicksum(x[i, j] for j in range(m)) == 1, f"item_{i}")
    
    for j in range(m):
        model.addConstr(
            quicksum(volumes[i] * x[i, j] for i in range(n)) <= capacities[j] * y[j],
            f"capacity_{j}"
        )
    
    # Optimize
    model.optimize()
    
    if model.status == GRB.OPTIMAL:
        # Extract solution
        bins_used = sum(1 for j in range(m) if y[j].X > 0.5)
        solution = np.zeros((n, m))
        for i in range(n):
            for j in range(m):
                if x[i, j].X > 0.5:
                    solution[i, j] = 1
        
        return solution, bins_used, model.objVal, model.Runtime
    else:
        return None, None, None, model.Runtime

# Visualization function
def plot_binpacking_solution(volumes, capacities, solution, title, save_to):
    plt.figure(figsize=(12, 8))
    n, m = solution.shape
    x = solution
    y = [1 if np.any(x[:, j] > 0.5) else 0 for j in range(m)]
    colors = plt.cm.Set3(np.linspace(0, 1, m))

    for j in range(m):
        if y[j] > 0.5:
            bin_items = [i for i in range(n) if x[i, j] > 0.5]
            bin_volumes = [volumes[i] for i in bin_items]
            
            # Create stacked bar for this bin
            bottom = 0
            for i, vol in enumerate(bin_volumes):
                plt.bar(j, vol, bottom=bottom, color=colors[j], alpha=0.7)
                plt.text(j, bottom + vol/2, f"Item {bin_items[i]+1}\n({vol})", 
                        ha='center', va='center', fontsize=8)
                bottom += vol
        # Draw bin capacity line
        plt.hlines(capacities[j], j-0.4, j+0.4, colors='red', linestyles='dashed')
        plt.text(j, capacities[j]+1, f"{capacities[j]}", 
                    ha='center', va='bottom', color='red', fontsize=8)
    plt.xlabel('Bin Number')
    plt.ylabel('Volume')
    plt.title(title)
    plt.legend()
    plt.xticks(range(m), [f'{i+1}' for i in range(m)])
    plt.savefig(save_to)
    plt.show()

# Save results function
def save_binpacking_solution(instance_name, num_items, num_bins, bins_used, runtime, filename):
    try:
        workbook = pxl.load_workbook(filename)
        sheet = workbook.active
    except FileNotFoundError:
        workbook = pxl.Workbook()
        sheet = workbook.active
        headers = ["Instance_Name", "Num_Items", "Num_Bins", "Bins_Used", "Runtime"]
        sheet.append(headers)

    row_data = [instance_name, num_items, num_bins, bins_used, f"{runtime:.3f}"]
    sheet.append(row_data)
    workbook.save(filename)

# Main execution
if os.path.exists("binpacking_random_solutions.xlsx"):
    os.remove("binpacking_random_solutions.xlsx")

seed = 42
num_items_sizes = [20, 50, 100, 200]
# Number of bins should be the quarter of number of items
num_bins_sizes = [max(1, n // 4) for n in num_items_sizes]
# print(num_items_sizes)
# print(num_bins_sizes)
for num_items, num_bins in zip(num_items_sizes, num_bins_sizes):
    volumes, capacities = generate_binpacking_instance(num_items, num_bins, seed, capacity_ratio=5)
    # print(f"Generated instance with {num_items} items and {num_bins} bins.")
    # print(f"Item volumes: {volumes}")
    # print(f"Bin capacities: {capacities}")
    solution, bins_used, obj_val, runtime = solve_binpacking(volumes, capacities, display=True)
    if solution is not None:
        print(f"Solved instance with {num_items} items and {num_bins} bins.")
        print(f"Optimal number of bins used: {obj_val}, Runtime: {runtime:.3f} seconds")
        instance_name = f"binpacking_{num_items}_items_{num_bins}_bins"
        save_binpacking_solution(instance_name, num_items, num_bins, bins_used, runtime, "binpacking_random_solutions.xlsx")
        if num_bins <= 25:  # Only plot small instances
            plot_binpacking_solution(volumes, capacities, solution, f"Bin Packing Solution for {num_items} items and {num_bins} bins", f"binpacking_solution_{num_items}_{num_bins}.png")
        else:
            # Print solution
            print("Solution:")
            print(f"Optimal number of bins: {obj_val}")
            for j in range(num_bins):
                if np.any(solution[:, j] > 0.5):
                    bin_items = [i for i in range(num_items) if solution[i, j] > 0.5]
                    bin_volume = sum(volumes[i] for i in bin_items)
                    print(f"  Bin {j+1} -> Volume {bin_volume}, Items {[i+1 for i in bin_items]}")
    else:
        print(f"Failed to solve instance with {num_items} items and {num_bins} bins. Runtime: {runtime:.3f} seconds")
        
```

## Improved Mathematical Formulation: Bin Packing with Minimization of Unused Space

Let:

- $N = \{1, 2, \dots, n\}$ be the set of items  
- $M = \{1, 2, \dots, m\}$ be the set of available bins  
- $v_i$ be the volume of item $i$  
- $C_j$ be the capacity of bin $j$  



### **Decision Variables**
- $x_{ij} \in \{0, 1\}$:  
  Equals 1 if item $i$ is assigned to bin $j$, 0 otherwise

- $y_j \in \{0, 1\}$:  
  Equals 1 if bin $j$ is used, 0 otherwise

- $e_j \in \mathbb{R}^{+}$:  
  Unused (empty) space in bin $j$



### **Objective Function**

We aim to **minimize the number of bins used** **and** the **total unused space**:

$$
\min Z = \alpha \sum_{j \in M} y_j + \beta \sum_{j \in M} e_j
$$

- $\alpha$: weight assigned to minimizing the number of bins (typically large, e.g., 1000)  
- $\beta$: weight assigned to minimizing unused space (typically small, e.g., 1)

This encourages solutions with fewer bins, and among those, with tighter packing.



### **Constraints**

1. **Item Assignment**:  
   Each item must be assigned to exactly one bin:

$$
\sum_{j \in M} x_{ij} = 1 \quad \forall i \in N
$$

2. **Bin Capacity**:  
   Total volume in each bin cannot exceed its capacity:

$$
\sum_{i \in N} v_i x_{ij} \leq C_j y_j \quad \forall j \in M
$$

3. **Empty Space Definition**:

$$
e_j = C_j y_j - \sum_{i \in N} v_i x_{ij} \quad \forall j \in M
$$

4. **Non-Negativity of Empty Space**:

$$
e_j \geq 0 \quad \forall j \in M
$$



### Remarks

- The objective function is **weighted**, and can be adapted:
  - Pure bin minimization: set $\beta = 0$
  - Pure space minimization given bins used: set $\alpha = 0$
  - Balanced trade-off: set $\alpha \gg \beta$ (e.g., $\alpha = 1000$, $\beta = 1$)

- An **ε-constraint method** can be used instead, where we minimize unused space under a fixed number of bins, or vice versa.


```python
# Function to solve the bin packing problem with improved formulation
def solve_binpacking_dual(volumes, capacities, display, alpha, beta):
    n = len(volumes)
    m = len(capacities)
    
    model = Model("random-bin-packing")
    
    if not display:
        model.setParam('OutputFlag', 0)
    
    # Create variables
    x = model.addVars(n, m, vtype=GRB.BINARY, name="x")
    y = model.addVars(m, vtype=GRB.BINARY, name="y")
    # e should be non-negative continuous variable representing empty space in each bin
    e = model.addVars(m, vtype=GRB.CONTINUOUS, name="e")

    # Set objective
    model.setObjective(alpha * quicksum(y[j] for j in range(m)) + beta * quicksum(e[j] for j in range(m)), GRB.MINIMIZE)

    # Add constraints
    for i in range(n):
        model.addConstr(quicksum(x[i, j] for j in range(m)) == 1, f"item_{i}")

    for j in range(m):
        model.addConstr(
            quicksum(volumes[i] * x[i, j] for i in range(n)) <= capacities[j] * y[j],
            f"capacity_{j}"
        )

    for j in range(m):
        model.addConstr(e[j] == capacities[j] * y[j] - quicksum(volumes[i] * x[i, j] for i in range(n)), f"empty_space_{j}")

    for j in range(m):
        model.addConstr(e[j] >= 0, f"nonneg_empty_{j}")
    # Optimize
    model.optimize()

    if model.status == GRB.OPTIMAL:
        # Extract solution
        bins_used = sum(1 for j in range(m) if y[j].X > 0.5)
        solution = np.zeros((n, m))
        for i in range(n):
            for j in range(m):
                if x[i, j].X > 0.5:
                    solution[i, j] = 1

        return solution, bins_used, model.objVal, model.Runtime
    else:
        return None, None, None, model.Runtime

if os.path.exists("binpacking_random_dual_solutions.xlsx"):
    os.remove("binpacking_random_dual_solutions.xlsx")
seed = 42
num_items_sizes = [20, 50, 100, 200]
# Number of bins should be the quarter of number of items
num_bins_sizes = [max(1, n // 4) for n in num_items_sizes]
# print(num_items_sizes)
# print(num_bins_sizes)
for num_items, num_bins in zip(num_items_sizes, num_bins_sizes):
    volumes, capacities = generate_binpacking_instance(num_items, num_bins, seed, capacity_ratio=5)
    print(f"Generated instance with {num_items} items and {num_bins} bins.")
    # print(f"Item volumes: {volumes}")
    # print(f"Bin capacities: {capacities}")
    solution, bins_used, obj_val, runtime = solve_binpacking_dual(volumes, capacities, display=True, alpha=1000, beta=1)
    if solution is not None:
        print(f"Solved instance with {num_items} items and {num_bins} bins.")
        print(f"Optimal number of bins used: {obj_val}, Runtime: {runtime:.3f} seconds")
        instance_name = f"binpacking_{num_items}_items_{num_bins}_bins"
        save_binpacking_solution(instance_name, num_items, num_bins, bins_used, runtime, "binpacking_random_dual_solutions.xlsx")
        if num_bins <= 25:  # Only plot small instances
            plot_binpacking_solution(volumes, capacities, solution, f"Bin Packing Solution for {num_items} items and {num_bins} bins", f"binpacking_dual_solution_{num_items}_{num_bins}.png")
        else:
            # Print solution
            print("Solution:")
            print(f"Optimal number of bins: {obj_val}")
            for j in range(num_bins):
                if np.any(solution[:, j] > 0.5):
                    bin_items = [i for i in range(num_items) if solution[i, j] > 0.5]
                    bin_volume = sum(volumes[i] for i in bin_items)
                    print(f"  Bin {j+1} -> Volume {bin_volume}, Items {[i+1 for i in bin_items]}")
    else:
        print(f"Failed to solve instance with {num_items} items and {num_bins} bins. Runtime: {runtime:.3f} seconds")
        
```
