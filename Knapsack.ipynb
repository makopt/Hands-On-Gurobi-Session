## Problem Definition
Given a **set $N$ of $n$ of items** $N=\{1, 2, \dots, n\}$, each with a **weight $w_i$** and a **value $v_i$**, and a **knapsack** with a **capacity $W$**. The **knapsack problem** aims to determine the **items to be selected** and placed in the knapsack such that the **total weight does not exceed $W$** and the **total value is maximized**.

---

## Mathematical Formulation
The 0/1 Knapsack Problem can be formulated as follows:
- **Decision Variables**:
  - $\forall i \in N, x_i = 1$ if item $i$ is included in the knapsack, $0$ otherwise.
- **Objective Function**:
  - Max Z=$\sum_{i=1}^{n} v_i x_i$
- **Constraint**:
  - $\sum_{i=1}^{n} w_i x_i \leq W$

## First Example (Solving a small manually created instance of the Knapsack Problem)

Let's solve a small manually created instance of the Knapsack Problem using Gurobi in Python.

### Overview

Thw following Python script is a straightforward, self-contained implementation for solving a **0-1 Knapsack Problem** using the Gurobi optimizer. Unlike the benchmarking suite, this code focuses on solving a single, manually defined problem instance with a fixed set of items. It serves as an educational example or a template for small-scale knapsack problems.

### Key Characteristics

*   **Manual Instance:** The problem data (values and weights for 10 items) is hard-coded into the script.
*   **Fixed Capacity:** The knapsack's capacity is set to approximately the half the total weight of all items (`sum(weights) / 2`).
*   **Basic Implementation:** It demonstrates the core steps of setting up and solving a Mixed-Integer Programming (MIP) model with Gurobi without the advanced features of the benchmarking suite (e.g., automated generation, scaling, or persistent logging).
*   **Visualization Included:** The solution is presented both textually and through a detailed matplotlib visualization.

### Code Workflow

#### 1. Data Preparation
The problem data is explicitly defined in the script:
```python
n = 10 # Number of items
values = [10, 5, 15, 7, 6, 18, 3, 12, 8, 14] # Value of each item
weights = [2, 3, 5, 7, 1, 4, 3, 6, 2, 5] # Weight of each item
capacity = np.sum(weights) / 2 # Knapsack capacity (50% of total weight)
```

#### 2. Gurobi Model Setup

- **Model Creation**: A new Gurobi model named `small-knapsack` is initialized.

- **Decision Variables**: Creates 10 binary variables (`x[0]` to `x[9]`), each representing whether an item is selected (1) or not (0).

- **Objective Function**: The goal is to maximize the total value of selected items. The objective is written out explicitly for each variable, which is feasible for a small $n=10$ but not scalable. 
```python
m.setObjective(
    10 * x[0] + 5 * x[1] + 15 * x[2] + 7 * x[3] + 6 * x[4] +
    18 * x[5] + 3 * x[6] + 12 * x[7] + 8 * x[8] + 14 * x[9],
    GRB.MAXIMIZE
)
```

- **Constraint**: Adds a single constraint ensuring the total weight of selected items does not exceed the predefined capacity. Like the objective, this is written explicitly.
```python
m.addConstr(
    2 * x[0] + 3 * x[1] + 5 * x[2] + 7 * x[3] + 1 * x[4] +
    4 * x[5] + 3 * x[6] + 6 * x[7] + 2 * x[8] + 5 * x[9] <= capacity,
    "Capacity"
)
```

#### 3. Model Solving and Output

- **Export**: The model is written to a file ('small-knapsack.lp') in LP format for inspection.

- **Optimization**: The `m.optimize()` command triggers the Gurobi solver to find the optimal solution.

- **Console Output**:

    - Prints the name and value of each selected variable (item).

    - Prints the total objective value (total value of the knapsack).

    - Calculates and prints the total weight of the selected items.

    - Prints the capacity and the solver's runtime.

#### 4. Solution Visualization
A bar chart is generated to provide a visual representation of the solution:

- **Blue Bars:** Show the weight of each item (on the y-axis).

- **Green Bars:** Show the value of items that were selected.

- **Red Bars:** Show the value of items that were not selected.

- **Annotations:** Each bar is annotated with its exact weight (`w=...`) and value (`v=...`).

- **Capacity Line:** A green dashed line indicates the knapsack's capacity constraint.

### Dependencies

- `gurobipy`: The optimization solver.

- `numpy`: Used for the sum function to calculate capacity.

- `matplotlib`: For generating the solution plot.

### Purpose and Use Case
This code is ideal for:

- **Learning:** Understanding the basic structure of a Gurobi MIP model for the knapsack problem.

- **Prototyping:** Quickly testing a specific, small problem instance.

- **Visualization:** Seeing the relationship between item weights, values, and the optimal selection in a clear plot.

### Limitations

- **Not Scalable:** The objective and constraint are hard-coded for exactly 10 items. Adding an 11th item would require manual code modification.

- **Single Instance:** Only solves one fixed problem. It lacks the automation to run multiple or generated instances.

- **No Result Logging:** Results are printed to the console and plotted but not saved to a file for future analysis.



````python
from gurobipy import GRB, Model
import numpy as np
import matplotlib.pyplot as plt

# Prepare data
n = 20  # number of items
values = [30, 35, 10, 20, 15, 5, 5, 40, 35, 25, 20, 20, 5, 25, 20, 10, 25, 45, 10, 5]
weights = [13, 17, 9, 10, 8, 7, 13, 18, 10, 7, 14, 4, 3, 12, 23, 4, 11, 13, 19, 27]
capacity = sum(weights) // 2  # set capacity to half of total weight

# Create a new model
m = Model("small-knapsack")

# Create variables
x = m.addVars(n, vtype=GRB.BINARY, name="x")

# Set objective
m.setObjective(values[0] * x[0] + values[1] * x[1] + values[2] * x[2]
               + values[3] * x[3] + values[4] * x[4] + values[5] * x[5]
               + values[6] * x[6] + values[7] * x[7] + values[8] * x[8]
               + values[9] * x[9] + values[10] * x[10] + values[11] * x[11]
               + values[12] * x[12] + values[13] * x[13] + values[14] * x[14]
               + values[15] * x[15] + values[16] * x[16] + values[17] * x[17]
               + values[18] * x[18] + values[19] * x[19],  GRB.MAXIMIZE)

# Add constraints
m.addConstr(weights[0] * x[0] + weights[1] * x[1] + weights[2] * x[2]
            + weights[3] * x[3] + weights[4] * x[4] + weights[5] * x[5]
            + weights[6] * x[6] + weights[7] * x[7] + weights[8] * x[8]
            + weights[9] * x[9] + weights[10] * x[10] + weights[11] * x[11]
            + weights[12] * x[12] + weights[13] * x[13] + weights[14] * x[14]
            + weights[15] * x[15] + weights[16] * x[16] + weights[17] * x[17]
            + weights[18] * x[18] + weights[19] * x[19] <= capacity, "capacity")

# Write model to a file (optional)
m.write("small-knapsack.lp")

# Optimize model
m.optimize()

# Print solution
for v in m.getVars():
    if v.x > 0.0001:  # only printing non-zero variables
        print(f"{v.varName}: {v.x}")    
print(f"Obj: {m.objVal}")
print(f"Total weight: {sum(weights[i] for i in range(n) if x[i].x > 0.5)}")
print(f"Capacity: {capacity}")
print(f"Runtime: {m.Runtime} seconds")

# Plot solution
# Create bar charts
plt.figure(figsize=(20, 10))
plt.bar(range(n), weights, color='b', alpha=0.6, label='Weights')
plt.bar(range(n), [values[i] if x[i].x > 0.5 else 0 for i in range(n)], color='g', alpha=0.6, label='Selected Values')
plt.bar(range(n), [values[i] if x[i].x <= 0.5 else 0 for i in range(n)], color='r', alpha=0.6, label='Not Selected Values')
# Annotate each bar with value and weight
for i in range(n):
    plt.text(i, weights[i] + 0.5, f"w={weights[i]}", ha='center', va='bottom', color='blue', fontsize=8)
    if x[i].x > 0.5:
        plt.text(i, values[i] + 0.5, f"v={values[i]}", ha='center', va='bottom', color='green', fontsize=8)
    else:
        plt.text(i, values[i] + 0.5, f"v={values[i]}", ha='center', va='bottom', color='red', fontsize=8)
# Annotate capacity line
plt.axhline(y=capacity, color='g', linestyle='--', label='Capacity')
# Annotate x-axis and y-axis
plt.xlabel('Item Index')    
plt.ylabel('Weight / Value')
# Add title and legend
plt.title('Knapsack Problem Solution (Small Manual Instance)')
plt.legend()
# Customize x-axis and y-axis
plt.xticks(range(0, n+1, 5))  # x-axis ticks at interval 5
plt.yticks(range(0, (int)(max(max(weights), max(values))) + 2, 5))  # y-axis ticks at interval 5
plt.show()
````

## Second Example (Solving a random generated instances of the Knapsack Problem)

### Overview

The following Python code is a comprehensive benchmarking and analysis suite for the **0-1 Knapsack Problem**. It systematically tests the performance of the Gurobi optimization solver on randomly generated, intentionally difficult knapsack instances of exponentially increasing size. Its primary goals are to:

1.  **Measure** solver runtime and performance scaling.
2.  **Visualize** the structure of problems and their solutions.
3.  **Log** all results persistently for further analysis.

### Components

#### 1. `generate_knapsack_instance(num_items, seed)`
Generates non-trivial, "difficult" problem instances.

*   **Controlled Randomness:** Uses a seed for reproducibility.
*   **Realistic Weight Distribution:** Weights are clustered around a `base_weight` that scales with `num_items`.
*   **Hard Items:** Creates a subset (~10%) of "very heavy" items to force complex trade-offs.
*   **Semi-Correlated Values:** Values are partially correlated with weights but include significant random noise.
*   **Tight Constraint:** Knapsack capacity is set to **25% of the total sum of all item weights**, making the problem highly constrained.

**Returns:** `weights, values, capacity`

#### 2. `solve_knapsack(weights, values, capacity)`
The core optimization engine using Gurobi.

*   **Model:** Formulates the problem as a Binary Integer Programming (BIP) model.
*   **Variables:** Creates a binary variable `x[i]` for each item (1 = include, 0 = exclude).
*   **Objective:** Maximizes the total value of selected items.
*   **Constraint:** Total weight of selected items must be <= capacity.
*   **Output:** Returns the solution selection, total value, total weight, and crucial **computation time**.

**Returns:** `solution, total_value, total_weight, runtime`

#### 3. `plot_knapsack_solution(weights, values, solution, capacity, title)`
Provides visual insight into the solver's decisions.

*   **Visualization:** A combined bar chart showing:
    *   **Blue Bars:** Weight of each item.
    *   **Green Bars:** Value of selected items.
    *   **Red Bars:** Value of non-selected items.
*   **Annotation:** Each bar is annotated with its exact weight (`w`) and value (`v`).
*   **Smart Plotting:** Only called for instances with `n <= 200` to avoid clutter.

#### 4. `save_knapsack_solution(instance_name, total_weight, total_value, runtime, filename, capacity)`
Handles structured storage of results.

*   **Persistence:** Appends results to an Excel file or creates a new one with headers if it doesn't exist.
*   **Data Logged:** Instance name, capacity, total weight, total value, and runtime for each experiment.

#### 5. Main Execution Loop
Orchestrates the benchmarking process.

*   **Clean Start:** Removes any existing results file to ensure a fresh benchmark.
*   **Exponential Scaling:** Tests problem sizes from **20 to 1,000,000 items** to stress-test the solver.
*   **Pipeline:** For each size `n`:
    1.  **Generate** a problem instance.
    2.  **Solve** it with Gurobi.
    3.  **Print** the solution to the console.
    4.  **Save** the results to Excel.
    5.  **Plot** the solution (if `n <= 200`).

### Key Features & Highlights

*   **Benchmarking Focus:** A framework for analyzing solver performance, not just a single solution.
*   **Algorithmic Hardness:** Instance generation is designed to create complex problems that test the solver's true IP capabilities.
*   **Resource Awareness:** Efficiently manages resources by avoiding unnecessary visualizations for large `n`.
*   **Complete Workflow:** Encompasses the entire data science process: generation → processing → visualization → persisted results.
*   **Scalability Analysis:** The range of `num_items` allows for clear analysis of runtime vs. problem size, revealing the practical limits of an exact solver like Gurobi.

### Dependencies

*   `gurobipy`: The core optimization solver.
*   `numpy`: For numerical operations and random number generation.
*   `matplotlib`: For generating solution plots.
*   `openpyxl`: For saving results to Excel files.
*   `os`: For file operations.

### Usage

The code is self-executing. Running the script will:
1.  Delete the existing `knapsack_random_solutions.xlsx` file.
2.  Iterate through the defined list of problem sizes.
3.  For each size, generate, solve, log, and (if applicable) plot the knapsack problem.
4.  Output results to the console and save them to the Excel file.
```


````python
from gurobipy import GRB, Model, quicksum
import numpy as np
import matplotlib.pyplot as plt
import openpyxl as pxl
import os
# Function to generate a difficult knapsack problem instance
def generate_knapsack_instance(num_items, seed):
    # Set random seed for reproducibility
    np.random.seed(seed)
    # Generate weights clustered around a base value, with some large outliers
    base_weight = max(1, int(10 + num_items))
    weights = np.random.randint(base_weight, base_weight * 4, size=num_items)
    # Add a few very heavy items to make the instance harder
    heavy_items = np.random.choice(num_items, size=max(1, num_items // 10), replace=False)
    for idx in heavy_items:
        weights[idx] = np.random.randint(base_weight * 5, base_weight * 8)
    # Generate values that are not proportional to weights, but with some correlation and noise
    values = weights + np.random.randint(-base_weight, base_weight * 2, size=num_items)
    values = np.maximum(values, 1)  # Ensure all values are positive
    # Set capacity to be a fraction of the total weight (e.g., 25% of total weight)
    capacity = int(weights.sum() * 0.25)
    return weights, values, capacity

# Function to solve the knapsack problem using Gurobi
def solve_knapsack(weights, values, capacity, display):
    num_items = len(weights)
    # Create a new model
    model = Model("random-knapsack")
    # Create variables
    x = model.addVars(num_items, vtype=GRB.BINARY, name="x")
    # Set objective
    model.setObjective(quicksum(values[i] * x[i] for i in range(num_items)), GRB.MAXIMIZE)
    # Add capacity constraint
    model.addConstr(quicksum(weights[i] * x[i] for i in range(num_items)) <= capacity, "capacity")
    # Write model to a file (optional)
    model.write("random-knapsack.lp")
    # No output if not display
    if not display:
        model.setParam('OutputFlag', 0)
    # Optimize the model
    model.optimize()
    # If display is True, print solution
    if display:
        for v in model.getVars():
            if v.x > 0.0001:  # only printing non-zero variables
                print(f"{v.varName}: {v.x}")    
    print(f"Obj: {model.objVal}")
    print(f"Total weight: {sum(weights[i] for i in range(num_items) if x[i].x > 0.5)}")
    print(f"Capacity: {capacity}")
    print(f"Runtime: {model.Runtime} seconds")
    # Extract solution
    solution = [int(x[i].X) for i in range(num_items)]
    total_value = model.ObjVal
    total_weight = sum(weights[i] for i in range(num_items) if solution[i] == 1)
    return solution, total_value, total_weight, model.Runtime

def plot_knapsack_solution(weights, values, solution, capacity, title):
    num_items = len(weights)
    # Create bar charts
    plt.figure(figsize=(20, 10))
    plt.bar(range(num_items), weights, color='b', alpha=0.6, label='Weights')
    plt.bar(range(num_items), [values[i] if solution[i] == 1 else 0 for i in range(num_items)], color='g', alpha=0.6, label='Selected Values')
    plt.bar(range(num_items), [values[i] if solution[i] != 1 else 0 for i in range(num_items)], color='r', alpha=0.6, label='Not Selected Values')
    # Annotate each bar with value and weight
    for i in range(num_items):
        plt.text(i, weights[i] + 0.5, f"w={weights[i]}", ha='center', va='bottom', color='blue', fontsize=8)
        if solution[i] == 1:
            plt.text(i, values[i] + 0.5, f"v={values[i]}", ha='center', va='bottom', color='green', fontsize=8)
        else:
            plt.text(i, values[i] + 0.5, f"v={values[i]}", ha='center', va='bottom', color='red', fontsize=8)
    # Annotate capacity line
    # plt.axhline(y=capacity, color='orange', linestyle='--', label='Capacity')
    # Annotate x-axis and y-axis
    plt.xlabel('Item Index')    
    plt.ylabel('Weight / Value')
    # Add title and legend
    plt.title(title)
    plt.legend()
    # Customize x-axis and y-axis
    plt.xticks(range(0, num_items+1, 5))  # x-axis ticks at interval 5
    plt.yticks(range(0, (int)(max(max(weights), max(values))) + 2, 20))  # y-axis ticks at interval 5
    # plt.grid()
    plt.savefig(f"knapsack_solution_{num_items}_cap_{capacity}.png")
    plt.show()

# Save solution in xlsx format file without using pandas
# This function will be called after solving each instance
# Each instance result will added in the first available line
def save_knapsack_solution(instance_name, total_weight, total_value, runtime, filename, capacity):
    try:
        # Try to load existing workbook
        workbook = pxl.load_workbook(filename)
        sheet = workbook.active
    except FileNotFoundError:
        # If file does not exist, create a new workbook and add headers
        workbook = pxl.Workbook()
        sheet = workbook.active
        headers = []
        headers.extend(["Instance_Name", "Capacity", "Total_Weight", "Total_Value", "Runtime"])
        sheet.append(headers)
    # Prepare data to write
    row_data = []
    runtime = f"{runtime:.3f}"
    row_data.extend([instance_name, capacity, total_weight, total_value, runtime])
    # Write data to the first empty row
    sheet.append(row_data)
    # Save the workbook
    workbook.save(filename)

# Example usage

# If the result excel file exists already, remove it
if os.path.exists("knapsack_random_solutions.xlsx"):
    os.remove("knapsack_random_solutions.xlsx")
seed = 42
sizes = [20, 50, 100, 200, 500, 1000, 2000, 5000, 10000, 20000, 50000, 100000, 200000, 500000, 1000000]
license_limit = [2000, 2000]  # Gurobi (named-user/free) license limit: [variables, constraints]
for num_items in sizes:
    weights, values, capacity = generate_knapsack_instance(num_items, seed)
    # This knapsack model has one binary variable per item and a single capacity constraint
    num_vars = num_items
    num_constraints = 1
    if num_vars > license_limit[0] or num_constraints > license_limit[1]:
        print(f"Skipping instance with {num_items} items: {num_vars} variables / {num_constraints} constraints exceeds license limit of {license_limit}")
        continue
    print(f"\nSolving Knapsack Problem with {num_items} items and capacity {capacity} (Random Instance)")
    solution, total_value, total_weight, runtime = solve_knapsack(weights, values, capacity, display=True if num_items <= 200 else False)
    save_knapsack_solution(f"KP_Random_n_{num_items}", total_weight, total_value, runtime, "knapsack_random_solutions.xlsx", capacity)    
    if num_items <= 200:  # Only plot for smaller instances to avoid clutter
        plot_knapsack_solution(weights, values, solution, capacity, f"Knapsack Problem (n={num_items}, capacity={capacity}) Solution (Random Instance)")
````

## Third Example (Solving a prestored instances of the Knapsack Problem)

### Overview

The following Python code is a comprehensive benchmarking and analysis suite for the **0-1 Knapsack Problem**. It systematically tests the performance of the Gurobi optimization solver on a set of pre-stored, intentionally difficult knapsack instances of exponentially increasing size. Its primary goals are to: 

1.  **Measure** solver runtime and performance scaling.
2.  **Visualize** the structure of problems and their solutions.
3.  **Log** all results persistently for further analysis.

### Components

The program uses the already implemented functions in the previous section such as:
- `solve_knapsack(weights, values, capacity)`
- `plot_knapsack_solution(weights, values, solution, capacity, title)`
- `save_knapsack_solution(instance_name, total_weight, total_value, runtime, filename, capacity)`
  
Only one new function is implemented to load the pre-stored instances from an Excel file.

#### 1. `load_knapsack_instance(folder_path)`

Loads knapsack instances from a specified Excel file resident in the given folder path.

*   **Input:** Path to the Excel file containing knapsack instances.
*   **Output:** A list of tuples, each containing (weights, values, capacity) for a knapsack instance.
*   **Excel Structure:** Assumes the Excel file has columns for weights, values, and capacity.

**Returns:** list of tuples `(weights, values, capacity)`

#### 2. Main Execution Loop
Orchestrates the benchmarking process.
*   **Clean Start:** Removes any existing results file to ensure a fresh benchmark.
*   **Pre-stored Instances:** Loads knapsack instances from an Excel file located in the specified folder path.
*   **Pipeline:** For each loaded instance:
    1.  **Solve** it with Gurobi.
    2.  **Print** the solution to the console.
    3.  **Save** the results to Excel.
    4.  **Plot** the solution (if `n <= 200`).  
    5.  **Output** results to the console and save them to the Excel file.

### Key Features & Highlights

*  **Benchmarking Focus:** A framework for analyzing solver performance, not just a single solution.
*  **Algorithmic Hardness:** Instance generation is designed to create complex problems that test the solver's true IP capabilities.
*  **Resource Awareness:** Efficiently manages resources by avoiding unnecessary visualizations for large `n`.

### Dependencies

*   `gurobipy`: The core optimization solver.
*   `numpy`: For numerical operations.
*   `matplotlib`: For generating solution plots.
*   `openpyxl`: For saving results to Excel files.
*   `os`: For file operations.

### Usage

The code is self-executing. Running the script will:
1.  Delete the existing `knapsack_prestored_solutions.xlsx` file.
2.  Load knapsack instances from the specified Excel file.
3.  For each instance, solve, log, and (if applicable) plot the knapsack problem.
4.  Output results to the console and save them to the Excel file. 

````python
# Load pre-stored knapsack instances from a folder
def load_knapsack_instances(folder_path, batches_number, batch_size):
    instances = []
    # readsorted files in the folder
    count = 0 # to count number of instances loaded
    for filename in sorted(os.listdir(folder_path)): # read files in sorted order
        if filename.endswith(".xlsx"): # only read excel files
            count += 1 # increment count
            if count > batches_number * batch_size: # limit number of instances loaded
                break
            filepath = os.path.join(folder_path, filename) # construct full file path
            workbook = pxl.load_workbook(filepath) # load workbook
            sheet = workbook.active # use the active sheet
            values = [] # to store values
            weights = [] # to store weights
            capacity = None # to store capacity
            for row in sheet.iter_rows(min_row=2, values_only=True):  # Skip header row
                if row[0] is not None and row[1] is not None: # ensure both value and weight are present
                    values.append(int(row[0])) # assuming values are in first column
                    weights.append(int(row[1])) # assuming weights are in second column
                if row[2] is not None: # assuming capacity is in third column of the first data row
                    capacity = int(row[2]) # to store capacity
            instances.append((filename, values, weights, capacity)) # append instance as a tuple
    return instances # return instances

# Main execution for pre-stored instances
results_file = "knapsack_pre_stored_solutions.xlsx" # Result file to save solutions
folder_path = "kp_instances"  # Folder containing pre-stored instances in txt format

if os.path.exists(results_file): # If the result excel file exists already, remove it
    os.remove(results_file)

sizes = [1000, 2000, 5000, 10000, 20000, 50000, 100000, 200000, 500000, 1000000] # Example sizes for batches
batches_number = 2  # Set the number of batches to test (between 1 and 10)
batch_size = 100    # Number of instances in each batch
instance_labels = [] # Names of instances that were actually solved
instance_runtimes = [] # Runtime of each instance that was actually solved
license_limit = [2000, 2000]  # Gurobi (named-user/free) license limit: [variables, constraints]
print("\nLoading pre-stored knapsack instances from folder:", folder_path)
instances = load_knapsack_instances(folder_path, batches_number, batch_size) # Load instances
for idx, (instance, values, weights, capacity) in enumerate(instances): # Iterate over loaded instances
    num_items = len(values) # number of items
    # This knapsack model has one binary variable per item and a single capacity constraint
    num_vars = num_items
    num_constraints = 1
    if num_vars > license_limit[0] or num_constraints > license_limit[1]: # Check both variables and constraints
        print(f"Skipping instance {instance} with {num_items} items: {num_vars} variables / {num_constraints} constraints exceeds license limit of {license_limit}")
        continue # skip to next instance
    print(f"\nSolving Pre-stored Instance {idx+1}: {instance} with {num_items} items and capacity {capacity}")
    solution, total_value, total_weight, runtime = solve_knapsack(weights, values, capacity, display=True if num_items <= 200 else False) # Solve instance
    save_knapsack_solution(instance, total_weight, total_value, runtime, results_file, capacity) # Save solution
    if num_items <= 200:  # Only plot for smaller instances to avoid clutter
        plot_knapsack_solution(weights, values, solution, capacity, f"Knapsack Problem (n={num_items}, capacity={capacity}) Solution (Pre-stored Instance {idx+1})") # Plot solution
    instance_labels.append(instance) # keep track of solved instance name
    instance_runtimes.append(runtime) # keep track of solved instance runtime

# Plot the runtime consumed by each solved instance, with a horizontal line for the average runtime
plt.figure(figsize=(20, 10)) # set figure size
instance_positions = np.arange(len(instance_runtimes)) # integer positions for each instance
plt.plot(instance_positions, instance_runtimes, marker='o', color='steelblue', label='Instance Runtime') # runtime per instance
average_runtime = sum(instance_runtimes) / len(instance_runtimes) if instance_runtimes else 0 # overall average runtime
plt.axhline(y=average_runtime, color='red', linestyle='--', label=f'Average Runtime ({average_runtime:.3f}s)') # average runtime line
plt.xlabel('Instance') # Set x-label
plt.ylabel('Runtime (seconds)') # Set y-label
plt.title('Knapsack Problem Runtime per Pre-stored Instance') # Set title
plt.legend() # Add legend
plt.xticks(instance_positions, instance_labels, rotation=90) # Label x-ticks with instance names
plt.tight_layout() # Adjust layout
plt.savefig("knapsack_instance_runtimes.png") # Save plot
plt.show() # Show plot
````
